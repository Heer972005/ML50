```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

# Target
y = df["defaulted_12m"]

# Remove target
X = df.drop(columns=["defaulted_12m"])

# Drop loan_id because it is an identifier, not a meaningful predictor
X = X.drop(columns=["loan_id"])

# Drop applicant_name because it is an identifier/name rather than a useful
# numeric or categorical modelling feature for this baseline
X = X.drop(columns=["applicant_name"])

# Convert loan_date and account_closed_date into useful numeric date features
for col in ["loan_date", "account_closed_date"]:
    X[col] = pd.to_datetime(X[col], errors="coerce")

    X[col + "_year"] = X[col].dt.year
    X[col + "_month"] = X[col].dt.month
    X[col + "_day"] = X[col].dt.day

    X = X.drop(columns=[col])

# Identify columns
numeric_features = X.select_dtypes(include=["number"]).columns
categorical_features = X.select_dtypes(include=["object", "category"]).columns

# Numeric preprocessing
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

# Combine preprocessing
preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])

# Baseline model
baseline_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])

# 5-fold stratified CV
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

auc_scores = cross_val_score(
    baseline_model,
    X,
    y,
    cv=cv,
    scoring="roc_auc"
)

print("5-fold AUC scores:", auc_scores)
print("Mean 5-fold AUC:", auc_scores.mean())
print("Standard deviation:", auc_scores.std())
```

```python
suspicious_features = [
    "collection_calls",
    "recovery_amount",
    "account_closed_date",
    "risk_flag",
    "late_but_repaid"
]

results = []

for feature in suspicious_features:

    # Start with the same X used for the baseline
    X_test = X.copy()

    # account_closed_date was converted into 3 columns
    if feature == "account_closed_date":
        cols_to_remove = [
            "account_closed_date_year",
            "account_closed_date_month",
            "account_closed_date_day"
        ]
    else:
        cols_to_remove = [feature]

    # Check that the columns exist
    missing_cols = [
        col for col in cols_to_remove
        if col not in X_test.columns
    ]

    if missing_cols:
        print(f"Skipping {feature}: columns not found -> {missing_cols}")
        continue

    X_test = X_test.drop(columns=cols_to_remove)

    # Identify column types again
    numeric_features_test = X_test.select_dtypes(
        include=["number"]
    ).columns

    categorical_features_test = X_test.select_dtypes(
        include=["object", "category"]
    ).columns

    preprocessor_test = ColumnTransformer([
        ("num",
         Pipeline([
             ("imputer", SimpleImputer(strategy="median")),
             ("scaler", StandardScaler())
         ]),
         numeric_features_test),

        ("cat",
         Pipeline([
             ("imputer", SimpleImputer(strategy="most_frequent")),
             ("onehot", OneHotEncoder(handle_unknown="ignore"))
         ]),
         categorical_features_test)
    ])

    model_test = Pipeline([
        ("preprocessor", preprocessor_test),
        ("model", LogisticRegression(max_iter=1000))
    ])

    scores = cross_val_score(
        model_test,
        X_test,
        y,
        cv=cv,
        scoring="roc_auc"
    )

    mean_auc = scores.mean()

    results.append({
        "Removed feature": feature,
        "Mean AUC": mean_auc,
        "Change from baseline": mean_auc - auc_scores.mean()
    })

results_df = pd.DataFrame(results)

display(results_df)
```

```python
# VERIFY near-duplicates
# Ignore loan_id because near-duplicates receive a new ID.
# Name formatting and monthly income can change.

# Normalize name
df["name_norm"] = (
    df["applicant_name"]
    .astype(str)
    .str.lower()
    .str.replace(r"[^a-z0-9]", "", regex=True)
)

# Normalize employment type
df["employment_norm"] = (
    df["employment_type"]
    .astype(str)
    .str.lower()
    .str.replace("-", "", regex=False)
    .str.replace(" ", "", regex=False)
)

# Income can be rounded in near-duplicates
df["income_norm"] = df["monthly_income"].round(-3)

# Fields that should remain the same for a true near-duplicate
check_cols = [
    "name_norm",
    "age",
    "employment_norm",
    "employment_years",
    "income_norm",
    "loan_amount",
    "loan_date",
    "credit_score",
    "region"
]

# Find candidate groups
groups = (
    df.groupby(check_cols, dropna=False)
      .size()
      .reset_index(name="count")
)

groups = groups[groups["count"] > 1]

print("Candidate groups:", len(groups))

# Get actual rows
candidates = df.merge(
    groups[check_cols],
    on=check_cols,
    how="inner"
)

# Remove exact duplicate rows first
candidates_no_exact = candidates.drop_duplicates(
    subset=[
        "applicant_name", "age", "employment_type",
        "employment_years", "monthly_income",
        "loan_amount", "loan_date", "credit_score",
        "region", "defaulted_12m"
    ]
)

print("Candidate rows after removing exact duplicates:",
      len(candidates_no_exact))

display(
    candidates_no_exact[
        [
            "loan_id",
            "applicant_name",
            "age",
            "employment_type",
            "employment_years",
            "monthly_income",
            "loan_amount",
            "loan_date",
            "credit_score",
            "region",
            "defaulted_12m"
        ]
    ].head(50)
)
```

```python
# FINAL VERIFICATION OF NEAR-DUPLICATES

# Fields that should be the same for a true near-duplicate
# loan_id is intentionally excluded because the generated near-duplicate gets a new ID.

verify_cols = [
    "name_norm",
    "age",
    "employment_norm",
    "employment_years",
    "income_norm",
    "loan_amount",
    "loan_date",
    "credit_score",
    "region"
]

# Find groups with the same normalized information
groups = (
    df.groupby(verify_cols, dropna=False)
      .size()
      .reset_index(name="count")
)

groups = groups[groups["count"] > 1]

# Get all candidate records
candidates = df.merge(
    groups[verify_cols],
    on=verify_cols,
    how="inner"
)

# Remove exact duplicates first
candidates = candidates.drop_duplicates()

# A true near-duplicate must have different loan IDs.
# Count rows that belong to groups with >1 UNIQUE loan_id.
verified_groups = (
    candidates.groupby(verify_cols, dropna=False)["loan_id"]
    .nunique()
    .reset_index(name="unique_loan_ids")
)

verified_groups = verified_groups[
    verified_groups["unique_loan_ids"] > 1
]

# Get the actual verified near-duplicate rows
verified = candidates.merge(
    verified_groups[verify_cols],
    on=verify_cols,
    how="inner"
)

print("Candidate groups:", len(groups))
print("Verified near-duplicate groups:", len(verified_groups))
print("Verified near-duplicate rows:", len(verified))

# Show examples
display(
    verified[
        [
            "loan_id",
            "applicant_name",
            "age",
            "employment_type",
            "employment_years",
            "monthly_income",
            "loan_amount",
            "loan_date",
            "credit_score",
            "region",
            "defaulted_12m"
        ]
    ].head(50)
)
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression

# --------------------------------------------------
# 1. Remove exact duplicate rows
# --------------------------------------------------

df_no_duplicates = df.drop_duplicates().copy()

print("Original dataset shape:", df.shape)
print("Dataset shape after removing duplicates:", df_no_duplicates.shape)
print("Exact duplicate rows removed:", len(df) - len(df_no_duplicates))


# --------------------------------------------------
# 2. Target
# --------------------------------------------------

y = df_no_duplicates["defaulted_12m"]


# --------------------------------------------------
# 3. Features
# --------------------------------------------------

X = df_no_duplicates.drop(columns=["defaulted_12m"])

# Remove identifiers
X = X.drop(columns=["loan_id"])
X = X.drop(columns=["applicant_name"])


# --------------------------------------------------
# 4. Convert date columns
# --------------------------------------------------

for col in ["loan_date", "account_closed_date"]:

    X[col] = pd.to_datetime(X[col], errors="coerce")

    X[col + "_year"] = X[col].dt.year
    X[col + "_month"] = X[col].dt.month
    X[col + "_day"] = X[col].dt.day

    X = X.drop(columns=[col])


# --------------------------------------------------
# 5. Identify numerical and categorical features
# --------------------------------------------------

numeric_features = X.select_dtypes(
    include=["number"]
).columns

categorical_features = X.select_dtypes(
    include=["object", "category"]
).columns


# --------------------------------------------------
# 6. Preprocessing
# --------------------------------------------------

numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_transformer, numeric_features),
    ("cat", categorical_transformer, categorical_features)
])


# --------------------------------------------------
# 7. Logistic Regression model
# --------------------------------------------------

model_no_duplicates = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])


# --------------------------------------------------
# 8. Same 5-fold CV as baseline
# --------------------------------------------------

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# --------------------------------------------------
# 9. Calculate AUC
# --------------------------------------------------

auc_scores_no_duplicates = cross_val_score(
    model_no_duplicates,
    X,
    y,
    cv=cv,
    scoring="roc_auc"
)


# --------------------------------------------------
# 10. Results
# --------------------------------------------------

print("5-fold AUC scores after removing duplicates:",
      auc_scores_no_duplicates)

print("Mean 5-fold AUC after removing duplicates:",
      auc_scores_no_duplicates.mean())

print("Standard deviation:",
      auc_scores_no_duplicates.std())
```

```python
# Find rows where defaulted_12m = 0 but other evidence looks suspicious

suspicious_labels = df[
    (df["defaulted_12m"] == 0) &
    (
        (df["collection_calls"] > 0) |
        (df["recovery_amount"] > 0) |
        (df["account_closed_date"].notna()) |
        (df["risk_flag"].astype(str).str.lower().isin(["high", "medium"]))
    )
].copy()

print("Number of suspicious label candidates:",
      len(suspicious_labels))

display(
    suspicious_labels[
        [
            "loan_id",
            "applicant_name",
            "collection_calls",
            "recovery_amount",
            "account_closed_date",
            "risk_flag",
            "late_but_repaid",
            "defaulted_12m"
        ]
    ].head(20)
)


suspicious_labels["evidence_count"] = (
    (suspicious_labels["collection_calls"] > 0).astype(int) +
    (suspicious_labels["recovery_amount"] > 0).astype(int) +
    (suspicious_labels["account_closed_date"].notna()).astype(int) +
    (
        suspicious_labels["risk_flag"]
        .astype(str)
        .str.lower()
        .isin(["high", "medium"])
    ).astype(int)
)

# Strongest candidates first
suspicious_labels = suspicious_labels.sort_values(
    "evidence_count",
    ascending=False
)

display(
    suspicious_labels[
        [
            "loan_id",
            "applicant_name",
            "collection_calls",
            "recovery_amount",
            "account_closed_date",
            "risk_flag",
            "late_but_repaid",
            "defaulted_12m",
            "evidence_count"
        ]
    ].head(20)
)

```
